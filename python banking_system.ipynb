```python
class Bank:
    def __init__(self):
        self.users = {}

    def create_account(self):
        acc_no = input("Enter Account Number: ")
        if acc_no in self.users:
            print("Account Already Exists!")
            return

        name = input("Enter Full Name: ")
        try:
            age = int(input("Enter Age: "))
            balance = float(input("Enter Initial Deposit: "))
        except ValueError:
            print("Invalid numeric input! Account creation aborted.")
            return

        phone = input("Enter Phone Number: ")
        password = input("Enter Password: ")

        self.users[acc_no] = {
            "name": name,
            "age": age,
            "phone": phone,
            "password": password,
            "balance": balance
        }
        print("Account created successfully!")

    def login(self):
        acc_no = input("Enter Account Number: ")
        password = input("Enter Password: ")

        if acc_no in self.users:
            if self.users[acc_no]["password"] == password:
                print(f"\nLogin Successful! Welcome, {self.users[acc_no]['name']}.")
                self.user_menu(acc_no)
            else:
                print("Wrong Password!")
        else:
            print("Account Not Found!")

    def deposit(self, acc_no):
        try:
            amount = float(input("Enter Deposit Amount: "))
        except ValueError:
            print("Invalid Amount!")
            return

        if amount > 0:
            self.users[acc_no]["balance"] += amount
            print("Deposit Successful!")
            print("Current Balance:", self.users[acc_no]["balance"])
        else:
            print("Invalid Amount!")

    def withdraw(self, acc_no):
        try:
            amount = float(input("Enter Withdrawal Amount: "))
        except ValueError:
            print("Invalid Amount!")
            return

        if amount > 0:
            if self.users[acc_no]["balance"] >= amount:
                self.users[acc_no]["balance"] -= amount
                print("Withdrawal Successful!")
                print("Remaining Balance:", self.users[acc_no]["balance"])
            else:
                print("Insufficient Balance!")
        else:
            print("Invalid Amount!")

    def check_balance(self, acc_no):
        print("Available Balance:", self.users[acc_no]["balance"])

    def check_profile(self, acc_no):
        user = self.users[acc_no]
        print("\n--- Account Profile ---")
        print(f"Account Number : {acc_no}")
        print(f"Name           : {user['name']}")
        print(f"Age            : {user['age']}")
        print(f"Phone          : {user['phone']}")
        print(f"Balance        : {user['balance']}")

    def user_menu(self, acc_no):
        while True:
            print("\n>>>>>> User Menu <<<<<<")
            print("1. Deposit")
            print("2. Withdraw")
            print("3. Check Balance")
            print("4. Check Profile")
            print("5. Logout")

            choice = input("Enter choice (1 to 5): ")

            if choice == "1":
                self.deposit(acc_no)
            elif choice == "2":
                self.withdraw(acc_no)
            elif choice == "3":
                self.check_balance(acc_no)
            elif choice == "4":
                self.check_profile(acc_no)
            elif choice == "5":
                print("Logged out successfully!")
                break
            else:
                print("Invalid Choice!")

    def run(self):
        while True:
            print("\n====== BANKING SYSTEM ======")
            print("1. Create Account")
            print("2. Login")
            print("3. Exit")
            choice = input("Enter choice (1 to 3): ")

            if choice == "1":
                self.create_account()
            elif choice == "2":
                self.login()
            elif choice == "3":
                print("Thank you for using the Banking System!")
                break
            else:
                print("Invalid Choice!")


if __name__ == "__main__":
    bank_app = Bank()
    bank_app.run()
```
